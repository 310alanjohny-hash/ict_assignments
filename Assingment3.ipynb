1. Write a Python function to fi nd the Max of three numbers


```python
a = 10
b = 5
c = 12

if a >= b and a >= c:
    largest = a
elif b >= a and b >= c:
    largest = b
else:
    largest = c

print("The largest number is:", largest)
```
